## 3.6 Approximations of Wavelet and Scaling Functions

### 3.6.1 Discrete Wavelet and Scaling Functions

```python
# import pywavelet
import pywt

# Create a Wavelet object for Daubechies wavelet with 8 vanishing moments ('db8')
wavelet = pywt.Wavelet('db8')

# Compute the scaling function (phi), wavelet function (psi), and corresponding time points (x)
# `level=3` defines the number of resolution levels to approximate the functions
phi, psi, x = wavelet.wavefun(level=1)

# print the phi and psi
print('db8 Scaling Function (φ):\n', phi)
print('db8 Wavelet Function (ψ):\n', psi)
```

### 3.6.2 Visualize the the functions

```python
# Import the PyWavelets library for wavelet functions
import pywt

# Import matplotlib for plotting
import matplotlib.pyplot as plt

# Create a Wavelet object for Daubechies wavelet with 8 vanishing moments ('db8')
wavelet = pywt.Wavelet('db8')

# Compute the scaling function (phi), wavelet function (psi), and corresponding time points (x)
# `level=1` defines the number of resolution levels to approximate the functions
phi, psi, x = wavelet.wavefun(level=1)

# Create a figure window with size 10x4 inches
plt.figure(figsize=(10, 4))

# Plot the scaling function φ(t)
plt.subplot(1, 2, 1)  # 1 row, 2 columns, 1st subplot
plt.plot(x, phi)
plt.title('Scaling Function $\phi(t)$ - db8')
plt.grid(True)

# Plot the wavelet function ψ(t)
plt.subplot(1, 2, 2)  # 1 row, 2 columns, 2nd subplot
plt.plot(x, psi)
plt.title('Wavelet Function $\psi(t)$ - db8')
plt.grid(True)

# Save the plot
plt.savefig("./output/3.1_scale_wavelet_function_db8.png", dpi=300)

# Show both subplots in the same figure
plt.show()
```

### 3.6.3 Biorthogonal Wavelets


```python
import pywt
import matplotlib.pyplot as plt

# Create the biorthogonal wavelet object
biwavelet = pywt.Wavelet('bior3.5')

# Get wavelet and scaling functions at level 5
phi_d, psi_d, phi_r, psi_r, x = biwavelet.wavefun(level=5)

# Set up a 2x2 grid of subplots
fig, axs = plt.subplots(2, 2, figsize=(12, 6))

# Plot decomposition wavelet ψ_d(t)
axs[0, 0].plot(x, psi_d)
axs[0, 0].set_title("Decomposition Wavelet Function ($\psi_d(t)$)")
axs[0, 0].set_xlabel("t")
axs[0, 0].grid(True)

# Plot reconstruction wavelet ψ_r(t)
axs[0, 1].plot(x, psi_r)
axs[0, 1].set_title("Reconstruction Wavelet Function ($\psi_r(t)$)")
axs[0, 1].set_xlabel("t")
axs[0, 1].grid(True)

# Plot decomposition scaling φ_d(t)
axs[1, 0].plot(x, phi_d)
axs[1, 0].set_title("Decomposition Scaling Function ($\phi_d(t)$)")
axs[1, 0].set_xlabel("t")
axs[1, 0].grid(True)

# Plot reconstruction scaling φ_r(t)
axs[1, 1].plot(x, phi_r)
axs[1, 1].set_title("Reconstruction Scaling Function ($\phi_r(t)$)")
axs[1, 1].set_xlabel("t")
axs[1, 1].grid(True)

# Adjust layout
plt.tight_layout()

# Save the figure
plt.savefig("/output/3.2_bior3.5_wavefun.png", dpi=300)

# Show the plot
plt.show()
```

### 3.6.4 Continuous Wavelet Functions

```python
import pywt
import matplotlib.pyplot as plt

# Create a ContinuousWavelet object with 'gaus2' (Gaussian wavelet, order 2)
wavelet = pywt.ContinuousWavelet('gaus2')

# Compute the wavelet function ψ(t) and its domain x at level 5
# Note: Continuous wavelets do not have a scaling function φ(t)
psi, x = wavelet.wavefun(level=5)

# Plot the continuous wavelet function ψ(t)
plt.figure(figsize=(8, 4))
plt.plot(x, psi, label="$\psi(t)$ - gaus2", color='navy')
plt.title("Continuous Wavelet Function ($\psi(t)$) - gaus2")
plt.xlabel("t")
plt.ylabel("Amplitude")
plt.grid(True)
plt.legend()

# Save the plot as a PNG file
plt.tight_layout()
plt.savefig("./output/3.3_gaus2_wavelet_function.png", dpi=300)

# Display the plot
plt.show()
```

## 3.7 Basic 1D Wavelet Decomposition

```python
import numpy as np
import matplotlib.pyplot as plt
import pywt

# Generate noisy signal
t = np.linspace(0, 1, 1024)
signal = np.sin(30 * 2 * np.pi * t) + 0.5 * np.random.randn(1024)

# Perform one-level decomposition
coeffs = pywt.dwt(signal, 'db4')
cA, cD = coeffs  # Approximation and detail coefficients

# Plot
plt.figure(figsize=(12, 6))
plt.subplot(3, 1, 1)
plt.plot(signal)
plt.title("Original Signal")

plt.subplot(3, 1, 2)
plt.plot(cA)
plt.title("Approximation Coefficients")

plt.subplot(3, 1, 3)
plt.plot(cD)
plt.title("Detail Coefficients")
plt.tight_layout()
plt.savefig('./output/3.4_simple_sine_dec.png')
plt.show()
```

## 3.8 Multilevel Decomposition and Reconstruction

```python
coeffs = pywt.wavedec(signal, 'db4', level=4)
reconstructed = pywt.waverec(coeffs, 'db4')
```

## 3.9 Signal Denoising

```python
import numpy as np
import pywt
import pywt.data
import matplotlib.pyplot as plt

# Load demo ECG signal from PyWavelets
signal = pywt.data.ecg()

# Add Gaussian noise to simulate a real-world noisy signal
noisy = signal + np.random.normal(0, 0.5, size=signal.shape)

# Perform multi-level wavelet decomposition using Daubechies-6 (db6)
coeffs = pywt.wavedec(noisy, 'db6', level=4)

# Estimate noise level from the last detail coefficients
sigma = np.std(coeffs[-1])

# Calculate universal threshold (VisuShrink)
uthresh = sigma * np.sqrt(2 * np.log(len(noisy)))

# Apply soft thresholding to all coefficients
coeffs_thresh = [pywt.threshold(c, value=uthresh, mode='soft') for c in coeffs]

# Reconstruct the signal from thresholded coefficients
reconstructed = pywt.waverec(coeffs_thresh, 'db6')

# Plot original, noisy, and denoised signals
plt.figure(figsize=(10, 6))
plt.plot(signal, label='Original', linewidth=1.5)
plt.plot(noisy, alpha=0.5, label='Noisy', linestyle='--')
plt.plot(reconstructed, label='Denoised', linewidth=1.5)
plt.legend()
plt.title("ECG Signal Denoising with PyWavelets")
plt.xlabel("Sample Index")
plt.ylabel("Amplitude")
plt.grid(True)

# Save the figure
plt.tight_layout()
plt.savefig("./output/3.5_ecg_wavelet_denoising.png", dpi=300)

# Display the figure
plt.show()
```
